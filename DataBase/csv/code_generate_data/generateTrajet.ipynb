```python
import random
import time


records = 500
print ("Making %d records\n", records)
```

```python
import csv

fieldnames=[
'client_id1',
'client_id2',
'client_id3',
'chauffeurtaxi_id',
'origins_Latitude', 
'origins_Longitude',
'destination_Latitude',
'destination_Longitude',
'distance_Km',
'dureetrajet_Mn',
'cout_Dt',
'date',
'heure',
'date_heure',
'evaluation_client',
'evaluation_trajet',
'nb utilisation taxi par semaine'
]
writer = csv.DictWriter(open("Trajet.csv","w"),fieldnames=fieldnames)
```

```python
import random
import string
def str_time_prop(start, end, time_format, prop):
    stime = time.mktime(time.strptime(start, time_format))
    etime = time.mktime(time.strptime(end, time_format))
    ptime = stime + prop * (etime - stime)
    return time.strftime(time_format, time.localtime(ptime))
def random_date(start, end, prop):
    return str_time_prop(start, end, '%m/%d/%Y %I:%M %p', prop)

writer.writerow(dict(zip(fieldnames,fieldnames)))
for i in range(0,records):
    if(i<200):
        a=random.randint(0,431)
        b=random.randint(0,431) 
        c=random.randint(0,431)
        while(a==b or a==c or b==c):  
            a=random.randint(0,431)
            b=random.randint(0,431) 
            c=random.randint(0,431)
        
        t=random_date("01/09/2022 1:00 PM", "01/09/2023 6:00 PM", random.random())
        h=random.randint(0,5)
        if (h>4):
            f=12
        elif (h==4):
            f=10
        elif (h>3):
            f=8
        elif (h==3):
            f=6
        elif (h>2):
            f=4
        elif (h>=2):
            f=2
        else:
            f=1
        writer.writerow(dict([
                ('client_id1',str(a)),
                ('client_id2',str(b)),
                ('client_id3',str(c)),
                ('chauffeurtaxi_id',"T"+str(random.randint(0,438))),
                ('origins_Latitude',str(random.random()+36)),
                ('origins_Longitude',str(random.random()+36)),
                ('destination_Latitude',str(random.random()+36)),
                ('destination_Longitude',str(random.random()+36)),
                ('distance_Km',str(random.randint(5,20))),
                ('dureetrajet_Mn',str(random.randint(5,30))),
                ('cout_Dt',str(random.randint(5,20))),
                ('date',t[0:10]),
                ('heure',t[11:]),
                ('date_heure',t),
                ('evaluation_client',str(random.randint(3,5))),
                ('evaluation_trajet',str(h)),
                ('nb utilisation taxi par semaine',str(f)),
                ]))

    elif(i>200) and (i<400):
        a=random.randint(0,431)
        b=random.randint(0,431)
        c='vide'
        while(a==b):  
            a=random.randint(0,431)
            b=random.randint(0,431)
            
        t=random_date("01/09/2022 1:00 PM", "01/09/2023 6:00 PM", random.random())
        h=random.randint(0,5)
        if (h>4):
            f=12
        elif (h==4):
            f=10
        elif (h>3):
            f=8
        elif (h==3):
            f=6
        elif (h>2):
            f=4
        elif (h>=2):
            f=2
        else:
            f=1
        writer.writerow(dict([
                ('client_id1',str(a)),
                ('client_id2',str(b)),
                ('client_id3',str(c)),
                ('chauffeurtaxi_id',"T"+str(random.randint(0,438))),
                ('origins_Latitude',str(random.random()+36)),
                ('origins_Longitude',str(random.random()+36)),
                ('destination_Latitude',str(random.random()+36)),
                ('destination_Longitude',str(random.random()+36)),
                ('distance_Km',str(random.randint(5,20))),
                ('dureetrajet_Mn',str(random.randint(5,30))),
                ('cout_Dt',str(random.randint(5,20))),
                ('date',t[0:10]),
                ('heure',t[11:]),
                ('date_heure',t),
                ('evaluation_client',str(random.randint(3,5))),
                ('evaluation_trajet',str(h)),
                ('nb utilisation taxi par semaine',str(f)),
                ]))
    elif(i>400):
        a=random.randint(0,431)
        b='vide' 
        c='vide'
        t=random_date("01/09/2022 1:00 PM", "01/09/2023 6:00 PM", random.random())
        h=random.randint(0,5)
        if (h>4):
            f=12
        elif (h==4):
            f=10
        elif (h>3):
            f=8
        elif (h==3):
            f=6
        elif (h>2):
            f=4
        elif (h>=2):
            f=2
        else:
            f=1

        writer.writerow(dict([
                ('client_id1',str(a)),
                ('client_id2',str(b)),
                ('client_id3',str(c)),
                ('chauffeurtaxi_id',"T"+str(random.randint(0,438))),
                ('origins_Latitude',str(random.random()+36)),
                ('origins_Longitude',str(random.random()+36)),
                ('destination_Latitude',str(random.random()+36)),
                ('destination_Longitude',str(random.random()+36)),
                ('distance_Km',str(random.randint(5,20))),
                ('dureetrajet_Mn',str(random.randint(5,30))),
                ('cout_Dt',str(random.randint(5,20))),
                ('date',t[0:10]),
                ('heure',t[11:]),
                ('date_heure',t),
                ('evaluation_client',str(random.randint(0,5))),
                ('evaluation_trajet',str(h)),
                ('nb utilisation taxi par semaine',str(f)),
                ]))
print("fin")
```
