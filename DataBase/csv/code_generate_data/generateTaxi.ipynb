```python
import random
import string
def get_random_string(length):
        letters = string.ascii_lowercase
        result_str = ''.join(random.choice(letters) for i in range(length))
        return  result_str
def random_char(char_num):
           return ''.join(random.choice(string.ascii_letters) for _ in range(char_num))


writer.writerow(dict(zip(fieldnames,fieldnames)))
for i in range(0,records):
        writer.writerow(dict([
                ('id',"T"+str(i)),
                ('firstname',random.choice (Choice_name)),
                ('lastname', random.choice (Choice_lastname)),
                ('mobileNumber',str(random.randint(20000000,29999999))),
                ('mail',random_char(7)+"@gmail.com"),
                ('password',str(get_random_string(8))),
                ('modelVoiture', random.choice (Choice_voiture)),
                ('matriculeNumber',str(random.randint(120,214))+"Tunisie"+str(random.randint(1,99999))),
                ('cin',str(random.randint(10000000,19999999))),
                ]))
print('fin')
```
