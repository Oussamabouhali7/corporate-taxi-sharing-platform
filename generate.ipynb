```python
Type_of_request = ['Technical problem', 'Sanitary problem','Relationnal problem','Emergency', 'other']
Text_of_request = ['Rideau blocked at the hall 2009','Lift out of order','Heating broken in room','Discomfort in the hall',
'Water leak in 3rd floor toilet','Fight in rooftoop', 'Lamp not working on 2nd floor','Projection screen not working','Harassment between students',
'Overturned rubbish bin', 'Video projector not working in room 1019','Computer not working in room 2012','Blocked chair amphitheatre 1',
'Nose bleed','Student stumbles on stairs','Broken window in room 1025','Fire in room 1026','Mask not worn 1st floor','Drunk on campus',
'Door stuck in room 2021','Blocked window in room 1014','Broken lamp in room 1022','Blocked toilet','Vending machine broken 1st floor', 
'Insalubrious wall 4th floor']
Service_concerned =['Health assistance', 'Technicals', 'Maintenance service','Other','Security','Emergency']
State_of_execution =['0','1']
Location= [0,1,2,3,4]  
```

```python
def str_time_prop(start, end, time_format, prop):
    stime = time.mktime(time.strptime(start, time_format))
    etime = time.mktime(time.strptime(end, time_format))
    ptime = stime + prop * (etime - stime)
    return time.strftime(time_format, time.localtime(ptime))


def random_date(start, end, prop):
    return str_time_prop(start, end, '%m/%d/%Y %I:%M %p', prop)
```

```python
writer.writerow(dict(zip(fieldnames,fieldnames)))
for i in range(0,records):
    writer.writerow(dict([
    ('Id',str(random.randint(100001,199999))),
    ('DTR',random_date("1/1/2008 1:30 PM", "1/1/2009 4:50 AM", random.random())),
    ('DTHR', random_date("1/1/2008 1:30 PM", "1/1/2009 4:50 AM", random.random())),
    ('Type', random.choice (Type_of_request)),
    ('Text', random.choice(Text_of_request)),
    ('Location', random.choice (Location)),
    ('Service', random.choice (Service_concerned)),
    ('State',random.choice (State_of_execution))
    ]))
        
                          
print ("fin")
```

```python
import random
def randomTimeRange():
    print('Get a random time between two times')
    start_time_input = '10:00:00'
    end_time_input = '19:00:43'

    start_time = start_time_input.split(':')
    end_time = end_time_input.split(':')

    start_hour = start_time[0]
    start_minute = start_time[1]
    start_seconds = start_time[2]

    end_hour = end_time[0]
    end_minute = end_time[1]
    end_seconds = end_time[2]

    # Get maximum end time for randrange
    if end_hour == '19' and end_minute != '00':
        max_hour = 19 + 1
    else:
        max_hour = start_hour

    if start_minute > end_minute:
        minutes = random.randrange(int(end_minute), int(start_minute))
    elif start_minute < end_minute:
        minutes = random.randrange(int(start_minute), int(end_minute))

    if start_hour == end_hour:
        hours = start_hour
    elif start_hour != end_hour:
        hours = random.randrange(int(start_hour), int(max_hour))

    if str(hours) == str(end_hour):
        minutes = random.randrange(int(start_minute), int(end_minute))
    else:
        minutes = random.randrange(0, 59)

    if start_seconds == end_seconds:
        seconds = start_seconds
    elif start_seconds > end_seconds:
        seconds = random.randrange(int(start_seconds), int(59))
    elif start_seconds < end_seconds:
        seconds = random.randrange(int(start_seconds), int(end_seconds))

    h = int(hours)
    m = int(minutes)
    s = int(seconds)

    return f"{h:02d}" + ':' + f"{m:02d}" + ':' + f"{s:02d}"

print(randomTimeRange())
```
